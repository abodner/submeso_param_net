```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm

r2_init = np.load('/orcd/data/abodner/001/greene/scratch/wb_init/r2_cnn_wb_init.npy')
```

```python
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(5, 4), dpi=600)
plt.rcParams.update({'font.size': 8})
fig.tight_layout(pad=10.0)

color = cm.cool_r(np.linspace(0, 1, 10)) 

legend_list = ['$U(-10^{-4}, 10^{-4})$', '$U(-10^{-3}, 10^{-3})$ ', '$U(-10^{-2}, 10^{-2})$',
               '$U(-5\cdot 10^{-2}, 5\cdot 10^{-2})$', '$U(-7\cdot 10^{-2}, 7\cdot 10^{-2})$', '$U(-10^{-1}, 10^{-1})$', '$U(-3\cdot 10^{-1}, 3\cdot 10^{-1})$', '$U(-5\cdot 10^{-1}, 5\cdot 10^{-1})$', '$U(-7\cdot 10^{-1}, 7\cdot 10^{-1})$', 'U(-1,1)']

for i_res in range(5):
    for i_init, c_loc in zip(range(10),color):
        if i_res == 4:
            plt.scatter(i_res,r2_init[i_res,i_init+1], s=50,color=c_loc, label = legend_list[i_init])
        else:
            plt.scatter(i_res,r2_init[i_res,i_init+1], s=50,color=c_loc)
    if i_res == 4:
        plt.scatter(i_res,r2_init[i_res,0], s=50, c='k', marker='*',label='Xavier')
    else:
        plt.scatter(i_res,r2_init[i_res,0], s=50, c='k', marker='*')
plt.legend(loc='center left', ncol=1, bbox_to_anchor=(1, 0.5),fontsize=6)
plt.ylim([0.3,1])
plt.xticks(range(5),['$1/12^o$','$1/8^o$','$1/4^o$','$1/2^o$','$1^o$']);
plt.ylabel('$R^2$ values')
plt.xlabel('Filter scales')

plt.savefig('r2_wanb_sensitivity.png')
```
